```python
#Find and remove loci that are fixed between the species 

import pandas as pd 

class_tab = pd.read_csv('Ult_class_tab.csv')
```

```python
class_tab = class_tab.set_index('Sample')
```

```python
class_tab_pk = class_tab.loc[class_tab['Species'] == 'Pk']

class_tab_pk = class_tab_pk.drop('Species', axis=1)
```

```python
class_tab_pv = class_tab.loc[class_tab['Species'] == 'Pv']

class_tab_pv = class_tab_pv.drop('Species', axis=1)
```

```python
pk_fixed = []
pk_fixed_alt = []

for column in class_tab_pk.columns: 
    value = class_tab_pk[column].sum()
    if value == 0: 
        pk_fixed.append(column)
    elif value == class_tab_pk.shape[0]: 
        pk_fixed_alt.append(column)
```

```python
pv_fixed = []
pv_fixed_alt = []

for column in class_tab_pv.columns: 
    value = class_tab_pv[column].sum()
    if value == class_tab_pv.shape[0]: 
        pv_fixed.append(column)
    elif value == 0: 
        pv_fixed_alt.append(column)
```

```python
pk_fixed_alt
```

```python
disc_cols_ref = []

for item in pk_fixed: 
    if item in pv_fixed: 
        disc_cols_ref.append(item)
```

```python
disc_cols_ref
```

```python
class_tab.shape
```

```python
class_tab_unfixed = class_tab.drop(disc_cols_ref, axis= 1)
```

```python
class_tab_unfixed.to_csv('Unifxed_class_tab.csv')
```
