```python
#Create train test splits 

import pandas as pd 

full_samps = pd.read_csv("Unifxed_class_tab.csv")
```

```python
full_samps = full_samps.set_index('Sample')
full_samps.shape
```

```python
from sklearn.model_selection import train_test_split

bar_train, bar_test= train_test_split(full_samps,
                                   random_state=104, 
                                   test_size=0.25, stratify= full_samps['Species'])
```

```python
bar_train.to_csv('train_barcode.csv')

bar_test.to_csv('test_barcode.csv')
```
